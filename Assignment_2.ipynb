# Loan Approval Decision Table Experiment

## 1. Initial Decision Table (Before Enhancements)

Let's define a simple decision table for loan approval based on two conditions: **Credit Score** and **Employment Status**.

| Rule | Credit Score (C1) | Employment Status (C2) | Loan Approved (Action) |
|---|---|---|---|
| R1 | >= 650 | Employed | Yes |
| R2 | < 650 | Employed | No |
| R3 | >= 650 | Unemployed | No |
| R4 | < 650 | Unemployed | No |

**Note**: For simplicity, if the credit score is below 650 or the applicant is unemployed, the loan is denied in this initial table.

## 2. Python Implementation of Initial Decision Table

```python
import pandas as pd

def evaluate_initial_decision_table(customer_record):
    credit_score = customer_record['Credit Score']
    employment_status = customer_record['Employment Status']

    if credit_score >= 650 and employment_status == 'Employed':
        return 'Yes'
    else:
        return 'No'
```

## 3. Test with Sample Customer Records (Initial Table)

```python
initial_customer_records = [
    {'Customer ID': 1, 'Credit Score': 700, 'Employment Status': 'Employed'},
    {'Customer ID': 2, 'Credit Score': 600, 'Employment Status': 'Employed'},
    {'Customer ID': 3, 'Credit Score': 720, 'Employment Status': 'Unemployed'},
    {'Customer ID': 4, 'Credit Score': 580, 'Employment Status': 'Unemployed'},
    {'Customer ID': 5, 'Credit Score': 680, 'Employment Status': 'Employed'},
    {'Customer ID': 6, 'Credit Score': 640, 'Employment Status': 'Employed'},
    {'Customer ID': 7, 'Credit Score': 710, 'Employment Status': 'Employed'},
    {'Customer ID': 8, 'Credit Score': 550, 'Employment Status': 'Unemployed'},
    {'Customer ID': 9, 'Credit Score': 690, 'Employment Status': 'Employed'},
    {'Customer ID': 10, 'Credit Score': 610, 'Employment Status': 'Employed'}
]

results_initial = []
for record in initial_customer_records:
    decision = evaluate_initial_decision_table(record)
    results_initial.append({**record, 'Loan Approved (Initial)': decision})

df_results_initial = pd.DataFrame(results_initial)
display(df_results_initial)
```

## 4. Enhanced Decision Table (with New Conditions)

Now, let's add two new conditions: **Customer Age (C3)** and **Debt-to-Income Ratio (C4)**. This will lead to a more comprehensive set of rules.

**New Conditions:**
- **Customer Age (C3):**
    - `>= 25` (Adult)
    - `< 25` (Young Adult)
- **Debt-to-Income Ratio (C4):**
    - `<= 0.36` (Low DTI)
    - `> 0.36` (High DTI)

**Updated Decision Table:**

| Rule | Credit Score (C1) | Employment Status (C2) | Customer Age (C3) | Debt-to-Income Ratio (C4) | Loan Approved (Action) |
|---|---|---|---|---|---|
| R1 | >= 650 | Employed | >= 25 | <= 0.36 | Yes |
| R2 | >= 650 | Employed | >= 25 | > 0.36 | No |
| R3 | >= 650 | Employed | < 25 | <= 0.36 | No (Requires co-signer/more scrutiny) |
| R4 | >= 650 | Employed | < 25 | > 0.36 | No |
| R5 | < 650 | Employed | Any | Any | No |
| R6 | Any | Unemployed | Any | Any | No |

**Note**: This enhanced table introduces more nuanced rules. For instance, a high credit score and employment might not be enough if the DTI is too high or the applicant is very young, indicating higher risk.

## 5. Python Implementation of Enhanced Decision Table

```python
def evaluate_enhanced_decision_table(customer_record):
    credit_score = customer_record['Credit Score']
    employment_status = customer_record['Employment Status']
    customer_age = customer_record['Customer Age']
    dti_ratio = customer_record['Debt-to-Income Ratio']

    # Rule R6: Unemployed always 'No'
    if employment_status == 'Unemployed':
        return 'No'

    # Rule R5: Credit Score < 650 always 'No' (if employed, otherwise caught by R6)
    if credit_score < 650:
        return 'No'

    # From here, we know: Credit Score >= 650 AND Employed

    # Rule R1: >= 650, Employed, >= 25, <= 0.36 -> Yes
    if customer_age >= 25 and dti_ratio <= 0.36:
        return 'Yes'

    # Rule R2: >= 650, Employed, >= 25, > 0.36 -> No
    if customer_age >= 25 and dti_ratio > 0.36:
        return 'No'

    # Rule R3: >= 650, Employed, < 25, <= 0.36 -> No
    if customer_age < 25 and dti_ratio <= 0.36:
        return 'No'

    # Rule R4: >= 650, Employed, < 25, > 0.36 -> No
    if customer_age < 25 and dti_ratio > 0.36:
        return 'No'

    return 'No' # Should not be reached if rules cover all cases, but as a fallback
```

## 6. Test with Sample Customer Records (Enhanced Table) and Compare Results

```python
enhanced_customer_records = [
    # Original 10 records with new data, ensuring diverse scenarios
    {'Customer ID': 1, 'Credit Score': 700, 'Employment Status': 'Employed', 'Customer Age': 35, 'Debt-to-Income Ratio': 0.25}, # R1 -> Yes
    {'Customer ID': 2, 'Credit Score': 600, 'Employment Status': 'Employed', 'Customer Age': 40, 'Debt-to-Income Ratio': 0.30}, # R5 -> No
    {'Customer ID': 3, 'Credit Score': 720, 'Employment Status': 'Unemployed', 'Customer Age': 28, 'Debt-to-Income Ratio': 0.20}, # R6 -> No
    {'Customer ID': 4, 'Credit Score': 580, 'Employment Status': 'Unemployed', 'Customer Age': 22, 'Debt-to-Income Ratio': 0.45}, # R6 -> No
    {'Customer ID': 5, 'Credit Score': 680, 'Employment Status': 'Employed', 'Customer Age': 50, 'Debt-to-Income Ratio': 0.50}, # R2 -> No (High DTI)
    {'Customer ID': 6, 'Credit Score': 640, 'Employment Status': 'Employed', 'Customer Age': 30, 'Debt-to-Income Ratio': 0.28}, # R5 -> No
    {'Customer ID': 7, 'Credit Score': 710, 'Employment Status': 'Employed', 'Customer Age': 22, 'Debt-to-Income Ratio': 0.30}, # R3 -> No (Young Age)
    {'Customer ID': 8, 'Credit Score': 550, 'Employment Status': 'Unemployed', 'Customer Age': 60, 'Debt-to-Income Ratio': 0.15}, # R6 -> No
    {'Customer ID': 9, 'Credit Score': 690, 'Employment Status': 'Employed', 'Customer Age': 27, 'Debt-to-Income Ratio': 0.35}, # R1 -> Yes
    {'Customer ID': 10, 'Credit Score': 610, 'Employment Status': 'Employed', 'Customer Age': 33, 'Debt-to-Income Ratio': 0.40},# R5 -> No
    # Additional records to highlight new rules explicitly
    {'Customer ID': 11, 'Credit Score': 750, 'Employment Status': 'Employed', 'Customer Age': 30, 'Debt-to-Income Ratio': 0.40}, # Initial Yes, Enhanced No (High DTI) - Rule R2
    {'Customer ID': 12, 'Credit Score': 700, 'Employment Status': 'Employed', 'Customer Age': 20, 'Debt-to-Income Ratio': 0.30}, # Initial Yes, Enhanced No (Young Age) - Rule R3
    {'Customer ID': 13, 'Credit Score': 730, 'Employment Status': 'Employed', 'Customer Age': 20, 'Debt-to-Income Ratio': 0.45}  # Initial Yes, Enhanced No (Young Age & High DTI) - Rule R4
]

results_enhanced = []
for record in enhanced_customer_records:
    initial_decision = evaluate_initial_decision_table(record)
    enhanced_decision = evaluate_enhanced_decision_table(record)
    results_enhanced.append({
        **record,
        'Loan Approved (Initial)': initial_decision,
        'Loan Approved (Enhanced)': enhanced_decision
    })

df_results_enhanced = pd.DataFrame(results_enhanced)
display(df_results_enhanced)
```

## 7. Report: Analysis of Enhanced Decision Table

### Additional Business Rules Introduced:

Two new conditions were introduced to the decision table:
1.  **Customer Age:** This condition categorizes applicants based on whether they are 25 years or older (`>= 25`) or younger than 25 (`< 25`). This allows for more granular risk assessment, acknowledging that younger applicants might pose a higher risk or require different considerations.
2.  **Debt-to-Income Ratio (DTI):** This condition assesses the applicant's DTI, categorizing it as `Low DTI (<= 0.36)` or `High DTI (> 0.36)`. A high DTI indicates a significant portion of an applicant's income is already committed to debt payments, which is a key indicator of their ability to manage additional debt.

### Impact on Loan Approval Decisions:

The introduction of these new conditions significantly refined the loan approval process, making it more stringent and comprehensive. Here's a summary of the impact observed in the sample data:

*   **Previously Approved, Now Denied:** Several records that would have been approved under the initial, simpler rules were denied under the enhanced rules:
    *   **Customer ID 5:** Initially 'Yes' (Credit Score 680, Employed), but changed to 'No' due to high DTI (0.50) (Rule R2).
    *   **Customer ID 7:** Initially 'Yes' (Credit Score 710, Employed), but changed to 'No' due to young age (22) (Rule R3).
    *   **Customer ID 11:** Initially 'Yes' (Credit Score 750, Employed), but changed to 'No' due to high DTI (0.40) (Rule R2).
    *   **Customer ID 12:** Initially 'Yes' (Credit Score 700, Employed), but changed to 'No' due to young age (20) (Rule R3).
    *   **Customer ID 13:** Initially 'Yes' (Credit Score 730, Employed), but changed to 'No' due to young age (20) and high DTI (0.45) (Rule R4).
*   **Consistent Denials:** Applicants with poor credit scores or unemployed status (e.g., Customer IDs 2, 3, 4, 6, 8, 10) remained denied under both tables, as these fundamental risk factors are still paramount.
*   **More Nuanced Approvals:** An applicant needs to meet all positive criteria (good credit, employed, reasonable age, low DTI) to be approved. This reduces the chances of approving high-risk individuals who might have appeared eligible under a simpler rule set.

### Advantages of Extending the Decision Table:

1.  **Increased Accuracy and Risk Mitigation:** By incorporating more relevant financial and demographic factors, the decision table can make more informed and accurate loan approval decisions, significantly reducing the risk of loan defaults.
2.  **Greater Specificity and Granularity:** The enhanced table allows for a more detailed breakdown of conditions and actions, addressing specific scenarios that simpler tables might overlook. This leads to more precise rule application.
3.  **Improved Business Logic Representation:** It better reflects complex business policies and regulations that often involve multiple interconnected criteria. The table explicitly lays out these rules, making them transparent and auditable.
4.  **Easier Maintenance and Updates:** While more complex, a well-structured decision table makes it relatively straightforward to add, modify, or remove rules as business needs or regulations change, without altering core application logic.
5.  **Enhanced Fairness and Consistency:** By defining clear rules for a wider range of conditions, the approval process becomes more objective and consistent across all applicants, minimizing human bias.
6.  **Better Decision Justification:** Each decision can be traced back to a specific rule, providing clear justification for approvals or denials to both internal stakeholders and applicants.

In conclusion, extending the decision table with Customer Age and Debt-to-Income Ratio leads to a more robust, accurate, and transparent loan approval system, aligning better with comprehensive risk management strategies.
